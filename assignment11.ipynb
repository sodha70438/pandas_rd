```python
import pandas as pd

jan_orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'customer': ['Rahul', 'Priya', 'Amit'],
    'food_item': ['Pizza', 'Burger', 'Biryani'],
    'amount': [250, 150, 300]
})

feb_orders = pd.DataFrame({
    'order_id': [104, 105, 106],
    'customer': ['Neha', 'Vijay', 'Riya'],
    'food_item': ['Pasta', 'Dosa', 'Sandwich'],
    'amount': [220, 100, 180]
})

result = pd.concat(
    [jan_orders, feb_orders],
    ignore_index=True
)

print(result)
```

```python
products = pd.DataFrame({
    'product_id': [1, 2, 3],
    'product_name': ['Laptop', 'Mobile', 'Headphones'],
    'category': ['Electronics', 'Electronics', 'Accessories']
})

ratings = pd.DataFrame({
    'rating': [4.5, 4.2, 3.9]
})

result = pd.concat(
    [products, ratings],
    axis=1
)

print(result)
```

```python
streams = pd.DataFrame({
    'user': [
        'Rahul', 'Rahul', 'Rahul', 'Rahul',
        'Priya', 'Priya', 'Priya', 'Priya',
        'Amit', 'Amit', 'Amit', 'Amit'
    ],
    'day': [
        'Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Monday', 'Tuesday', 'Wednesday', 'Thursday'
    ],
    'song_streams': [
        20, 25, 18, 30,
        15, 22, 28, 24,
        10, 17, 20, 16
    ]
})

result = streams.pivot(
    index='user',
    columns='day',
    values='song_streams'
)

print(result)
```

```python
orders = pd.DataFrame({
    'restaurant': [
        'Pizza House', 'Pizza House', 'Pizza House',
        'Burger Point', 'Burger Point', 'Burger Point',
        'Pizza House', 'Burger Point'
    ],
    'month': [
        'January', 'January', 'February',
        'January', 'February', 'February',
        'March', 'March'
    ],
    'total_amount': [
        500, 700, 600,
        300, 400, 500,
        800, 350
    ]
})

result = orders.pivot_table(
    index='restaurant',
    columns='month',
    values='total_amount',
    aggfunc='mean'
)

print(result)
```
