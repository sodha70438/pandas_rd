```python
import pandas as pd

df = pd.DataFrame({
    'username': [
        'CoolDude',
        'foodieQueen',
        'TravelKing',
        'STYLESTAR',
        'MusicLover'
    ]
})

df['username'] = df['username'].str.lower()

print(df)
```

```python



df = pd.read_csv(r'D:\python_rd\pandas_rd\spotify_songs.csv')

df['song'] = (
    df['song']
    .str.replace('Believer', '', regex=False)
    .str.strip()
)

print(df['song'])
```

```python


df = pd.DataFrame({
    'review': [
        'Fast delivery and good food',
        'Food was delicious',
        'Delivery was late',
        'Good service',
        'Quick DELIVERY'
    ]
})

result = df[
    df['review'].str.contains(
        'delivery',
        case=False,
        na=False
    )
]

print(result)
```

```python


categories = pd.Series([
    'Electronics/Mobiles/Smartphones',
    'Home/Kitchen/Appliances',
    'Fashion/Men/Shirts',
    'Beauty/Skincare/Cleansers'
])

df = pd.DataFrame({
    'Category': categories
})


df['MainCategory'] = df['Category'].str.split('/').str[0]

print(df)
```

```python

df = pd.DataFrame({
    'Title': [
        'How to Learn Python',
        'Best Travel Places',
        'Easy Cooking Tips'
    ]
})


df['Title'] = df['Title'].str.upper().str.replace(' ', '_', regex=False)

print(df)
```
