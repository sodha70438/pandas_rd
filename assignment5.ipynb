```python
import pandas as pd

# Sample IPL players dataset
data = {
    'player_name': ['Virat Kohli', 'MS Dhoni', 'Rohit Sharma',
                    'Jasprit Bumrah', 'Shubman Gill'],
    'team': ['RCB', 'CSK', 'MI', 'MI', 'GT'],
    'age': [37, None, 39, None, 26]
}

df = pd.DataFrame(data)
df.to_csv('ipl_players.csv', index=False)

df = pd.read_csv('ipl_players.csv')

print(df[df['age'].isna()])
```

```python
import pandas as pd

data = {
    'order_id': [101, 102, 103, 104, 105],
    'customer_name': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'delivery_rating': [5, None, 4, None, 3]
}

df = pd.DataFrame(data)


cleaned_df = df.dropna(subset=['delivery_rating'])


print(cleaned_df)
```

```python
import pandas as pd

data = {
    'product_name': ['Phone', 'Laptop', 'Headphones',
                     'Smartwatch', 'Tablet'],
    'rating': [4, 5, None, 3, None]
}

df = pd.DataFrame(data)


mean_rating = df['rating'].mean()

df['rating'] = df['rating'].fillna(mean_rating)

print(df)
```

```python
import pandas as pd

data = {
    'song_name': ['Song A', 'Song B', 'Song C', 'Song D', 'Song E'],
    'duration': [180, 240, None, 210, None]
}

df = pd.DataFrame(data)

median_duration = df['duration'].median()

df['duration'] = df['duration'].fillna(median_duration)

df.to_csv('spotify_cleaned.csv', index=False)

print(df)
print("\nCleaned CSV file saved successfully.")
```

```python
import pandas as pd
import numpy as np

data = {
    'username': ['user_amit', 'user_priya', 'user_rahul',
                 'user_neha', 'user_karan'],
    'followers': [1200, np.nan, 850, np.nan, 2000]
}

df = pd.DataFrame(data)

df['followers'] = df['followers'].fillna(0)


print(df)
```
