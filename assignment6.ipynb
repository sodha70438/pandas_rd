```python
import pandas as pd

data = {
    'product_id': [101, 102, 101, 103, 102, 101],
    'user_id': [1, 2, 1, 3, 2, 4],
    'rating': [5, 4, 3, 5, 2, 4],
    'review_text': [
        'Excellent', 'Good', 'Average',
        'Very good', 'Bad', 'Nice'
    ]
}

df = pd.DataFrame(data)

# Count duplicate rows based on product_id and user_id
duplicates_before = df.duplicated(
    subset=['product_id', 'user_id']
).sum()

print("Duplicate reviews before:", duplicates_before)

# Remove duplicate combinations, keeping first occurrence
df_cleaned = df.drop_duplicates(
    subset=['product_id', 'user_id'],
    keep='first'
)

duplicates_after = df_cleaned.duplicated(
    subset=['product_id', 'user_id']
).sum()

print("Duplicate reviews after:", duplicates_after)
print("\nCleaned DataFrame:")
print(df_cleaned)
```

```python
import pandas as pd

data = {
    'order_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'order_amount': [200, 250, 300, 220, 280, 260, 240, 2000]
}

df = pd.DataFrame(data)

Q1 = df['order_amount'].quantile(0.25)
Q3 = df['order_amount'].quantile(0.75)


IQR = Q3 - Q1


lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


outliers = df[
    (df['order_amount'] < lower_bound) |
    (df['order_amount'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nOutlier Order IDs:")
print(outliers['order_id'].tolist())
```

```python
import pandas as pd
from scipy.stats import zscore

def find_outliers(df, column='transaction_amount', threshold=2):
    """
    Return transactions whose absolute Z-score
    exceeds the specified threshold.
    """
    result = df.copy()

  
    if len(result) < 2:
        return result.iloc[0:0]

    result['z_score'] = zscore(result[column], nan_policy='omit')

    
    outliers = result[result['z_score'].abs() > threshold]

    return outliers


data = {
    'transaction_id': [201, 202, 203, 204, 205, 206, 207, 208],
    'transaction_amount': [100, 150, 120, 130, 110, 140, 125, 2000]
}

df = pd.DataFrame(data)


outliers = find_outliers(df)

print("Outlier transactions:")
print(outliers[['transaction_id', 'transaction_amount', 'z_score']])
```
