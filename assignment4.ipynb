```python
import pandas as pd

movies = pd.DataFrame({
    "title": [
        "Dangal", "3 Idiots", "Pathaan",
        "Jawan", "Bajrangi Bhaijaan", "Chak De India"
    ],
    "rating": [8.3, 8.4, 5.8, 7.0, 8.1, 8.2],
    "release_year": [2016, 2009, 2023, 2023, 2015, 2007]
})

result = movies.sort_values("rating", ascending=False)

print(result)
```

```python
import pandas as pd

ipl_teams = pd.DataFrame({
    "team": ["CSK", "MI", "RCB", "KKR", "GT"],
    "points": [14, 16, 12, 18, 10],
    "net_run_rate": [0.5, 0.8, -0.2, 1.1, -0.4]
})

result = ipl_teams.sort_index(ascending=False)

print(result)
# D:\python_rd\pandas_rd\pandas_assignment\assignment4.ipynb
```

```python
youtube = pd.DataFrame({
    "video_title": [
        "Cricket Highlights",
        "Music Video",
        "Tech Review"
    ],
    "views": [500000, 800000, 200000],
    "likes": [25000, 50000, 10000]
}, index=[10, 20, 30])

result = youtube.reset_index(drop=True)
print(result)
```

```python
restaurants = pd.DataFrame({
    "restaurant": [
        "Pizza Palace",
        "Food Hub",
        "Spice Garden",
        "Royal Thali"
    ],
    "rating": [4.5, 4.2, 3.8, 4.0]
}, index=[0, 1, 2, 3])

result = restaurants.reindex([1, 0, 2, 3])
print(result)
```

```python
products = pd.DataFrame({
    "product_name": [
        "Mouse", "Headphones", "Keyboard", "Smart Watch"
    ],
    "price": [500, 1500, 2000, 3000],
    "discount": [10, 25, 15, 40]
})

result = products.sort_values(
    "discount",
    ascending=False
)

result = result.reset_index(drop=True)
print(result)
```
