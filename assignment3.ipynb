```python
import pandas as pd

# Create IPL matches dataset
ipl = pd.DataFrame({
    "team1": ["CSK", "MI", "RCB", "GT", "RR",
              "KKR", "DC", "SRH", "PBKS", "LSG"],
    "team2": ["MI", "CSK", "GT", "RCB", "KKR",
              "RR", "SRH", "DC", "LSG", "PBKS"],
    "winner": ["CSK", "MI", "RCB", "GT", "RR",
               "KKR", "SRH", "DC", "PBKS", "LSG"],
    "total_runs": [195, 175, 210, 185, 225,
                   200, 160, 218, 190, 180]
})

# Filter matches where total_runs is between 180 and 220
result = ipl[
    (ipl["total_runs"] >= 180) &
    (ipl["total_runs"] <= 220)
]

print(result)
```
