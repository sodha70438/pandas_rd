```python

df = pd.DataFrame({
    "city": ["Mumbai", "Delhi", "Mumbai", "Delhi", "Mumbai"],
    "restaurant": ["A", "B", "A", "B", "C"],
    "order_amount": [500, 300, 700, 400, 600]
})

result = (
    df.groupby(["city", "restaurant"])
      .agg(avg_order_amount=("order_amount", "mean"))
)

print(result)
```

```python

df = pd.DataFrame({
    "category": ["Electronics", "Clothing", "Electronics",
                 "Clothing", "Electronics"],
    "region": ["North", "South", "North", "West", "South"],
    "sales": [50000, 20000, 70000, 30000, 40000],
    "units_sold": [100, 50, 150, 80, 90]
})

result = (
    df.groupby(["category", "region"])
      .agg(
          total_sales=("sales", "sum"),
          avg_units_sold=("units_sold", "mean")
      )
)

print(result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "region": ["North", "South", "North", "South", "East"],
    "revenue": [10000, 15000, 20000, 12000, 18000]
})

result = (
    df.groupby("region")
      .agg({
          "revenue": ["sum", "max", "min"]
      })
)

print(result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "artist": ["Artist A", "Artist B", "Artist C",
               "Artist D", "Artist E"],
    "genre": ["Pop", "Rock", "Pop", "Hip-Hop", "Rock"],
    "streams": [1000000, 800000, 1500000, 600000, 1200000]
})

result = (
    df.groupby("genre")
      .agg(
          total_streams=("streams", "sum"),
          average_streams=("streams", "mean")
      )
      .sort_values(by="total_streams", ascending=False)
)

print(result)
```
