```python
import pandas as pd

data = {
    "ProductName": ["Laptop", "Mobile", "Headphones", "Keyboard"],
    "Price": [50000, 20000, 1500, 800],
    "Qty": [2, 3, 5, 4]
}

df = pd.DataFrame(data)

df.to_csv("flipkart_sales.csv", index=False)

df = pd.read_csv("flipkart_sales.csv")

df["TotalValue"] = df["Price"] * df["Qty"]

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "PostID": [101, 102, 103, 104],
    "Likes": [1500, 2000000, 850, 12500],
    "Comments": [250, 1200, 2000000, 75]
})

# Helper function to format large numbers
def format_number(num):
    if num >= 1000000:
        return f"{num / 1000000:g}M"
    elif num >= 1000:
        return f"{num / 1000:g}K"
    else:
        return str(num)

df[["Likes", "Comments"]] = df[["Likes", "Comments"]].map(
    format_number
)

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Restaurant": ["Pizza Hut", "McDonald's", "Domino's", "Burger King"],
    "Price": [300, 150, 250, 200],
    "Qty": [2, 3, 2, 4],
    "DeliveryCharge": [40, 30, 25, 35]
})

# Calculate final order amount
df["FinalAmount"] = df.apply(
    lambda row: (row["Price"] * row["Qty"]) + row["DeliveryCharge"],
    axis=1
)

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "ProductName": ["T-Shirt", "Jeans", "Shoes", "Jacket"],
    "Price": [500, 1200, 2000, 2500],
    "Qty": [2, 1, 1, 2]
})

# Calculate GST at 5% of the total product cost
df["GST"] = (df["Price"] * df["Qty"]) * 0.05

print(df)
```
