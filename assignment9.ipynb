```python
import pandas as pd

dates = pd.Series([
    '15-01-2024',
    '22-03-2023',
    '10-07-2025',
    '05-11-2022',
    '28-09-2024'
])

dates = pd.to_datetime(dates, format='%d-%m-%Y')

print(dates)
```

```python

df = pd.DataFrame({
    'movie': ['Movie A', 'Movie B', 'Movie C'],
    'release_date': ['15-01-2020', '22-07-2022', '05-11-2024']
})

df['release_date'] = pd.to_datetime(
    df['release_date'],
    format='%d-%m-%Y'
)

df['year'] = df['release_date'].dt.year
df['month'] = df['release_date'].dt.month
df['day'] = df['release_date'].dt.day

print(df)
```

```python
 
today = pd.Timestamp.today().normalize()

dates = pd.date_range(
    start=today,
    periods=7,
    freq='D'
)

print(dates.tolist())
```

```python

df = pd.DataFrame({
    'user_id': [101, 102, 103, 104],
    'start_date': [
        '01-01-2025',
        '15-06-2025',
        '10-01-2026',
        '01-10-2026'
    ]
})

df['start_date'] = pd.to_datetime(
    df['start_date'],
    format='%d-%m-%Y'
)

today = pd.Timestamp.today().normalize()

df['tenure_days'] = (today - df['start_date']).dt.days

print(df)
```
