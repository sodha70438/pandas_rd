```python
import pandas as pd

users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Rahul', 'Priya', 'Amit', 'Neha', 'Vijay'],
    'city': ['Delhi', 'Mumbai', 'Pune', 'Jaipur', 'Surat']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 5, 2, 4],
    'order_amount': [500, 800, 300, 1200, 450, 600, 900]
})

result = pd.merge(users, orders, on='user_id', how='inner')

print(result)
```

```python
users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5, 6],
    'username': [
        'Rahul', 'Priya', 'Amit',
        'Neha', 'Vijay', 'Anjali'
    ],
    'city': [
        'Delhi', 'Mumbai', 'Pune',
        'Jaipur', 'Surat', 'Ahmedabad'
    ]
})

result = pd.merge(users, orders, on='user_id', how='left')

print(result)
```

```python
restaurants = pd.DataFrame({
    'restaurant_id': [1, 2, 3],
    'name': ['Spice Hub', 'Pizza House', 'Food Corner'],
    'city': ['Delhi', 'Mumbai', 'Pune']
})

zomato_reviews = pd.DataFrame({
    'review_id': [101, 102, 103],
    'restaurant_id': [1, 2, 4],
    'rating': [4.5, 3.8, 4.2],
    'reviewer_city': ['Delhi', 'Mumbai', 'Jaipur']
})

result = pd.merge(
    restaurants,
    zomato_reviews,
    on='restaurant_id',
    how='outer'
)

print(result)
```

```python
playlists = pd.DataFrame({
    'playlist_id': [101, 102, 103, 104],
    'user_id': [1, 2, 3, 4],
    'city': ['Delhi', 'Mumbai', 'Pune', 'Surat'],
    'genre': ['Pop', 'Rock', 'Jazz', 'Hip-Hop']
})

user_profiles = pd.DataFrame({
    'user_id': [1, 2, 3, 4],
    'username': ['Rahul', 'Priya', 'Amit', 'Neha'],
    'city': ['Delhi', 'Pune', 'Pune', 'Ahmedabad']
})

result = pd.merge(
    playlists,
    user_profiles,
    on=['user_id', 'city'],
    how='inner'
)

print(result)
```

```python
products = pd.DataFrame({
    'product_id': [1, 2, 3],
    'name': ['Laptop', 'Mobile', 'Headphones'],
    'category': ['Electronics', 'Electronics', 'Accessories']
})

ratings = pd.DataFrame({
    'rating_id': [101, 102, 103, 104],
    'product_id': [1, 2, 2, 4],
    'rating': [4.5, 4.0, 3.5, 5.0]
})

result = pd.merge(
    products,
    ratings,
    on='product_id',
    how='right'
)

print(result)
```
